```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Cargar el dataset de diamonds
df = sns.load_dataset('diamonds')
```

```python
df.info()
```

```python
df[['depth']].describe().T
```

```python
# Ver las primeras filas del dataset
fig,ax = plt.subplots(figsize=(8,4))
df.depth.plot.hist(ax=ax,bins=np.arange(55,70,1))
```

# Distribucion muestral

```python
df.depth
```

## 1. Seleccion de una muestra

```python
# Seleccionando 5 datos al azar de la columna depth
muestra = df.depth.sample(5)
muestra
```

## 2. Calcular la media muestral

```python
prom = muestra.mean()
prom
```

## 3. Almacenar el promedio muestral

```python
l_prom = []
l_prom
```

```python
l_prom.append(prom)
l_prom
```
